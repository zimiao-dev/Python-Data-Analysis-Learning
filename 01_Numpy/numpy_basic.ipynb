# ndarray
## 1. ndarray 的特性
### 多维性

```python
import numpy as np
```

```python
arr = np.array(5) # 创建一个 0 维 ndarray 数组
print(arr)
print('arr的维度: ', arr.ndim) # 数组的维度 number of dimensions
```

```python
arr = np.array([1, 2, 3]) # 创建一个 1 维的 ndarray 数组
print(arr)
print('arr的维度: ', arr.ndim) # 数组维度 number of dimensions
```

```python
arr = np.array([[1, 2, 3], [4, 5, 6]]) # 创建一个 2 维的 ndarray 数组
print(arr)
print('arr的维度: ', arr.ndim) # 数组维度 number of dimensions
```

### 同质性

```python
arr = np.array([1, 'hello']) # 数据类型不同，会被强制转换为相同的数据类型
print(arr)
```

```python
arr = np.array([1, 2.5]) # 数据类型不同，会被强制转换为相同的数据类型
print(arr)
```

## 2. ndarray 的属性

```python
arr = np.array(1)
print(arr)
print('数组的形状\n', arr.shape)
print('数组的维度', arr.ndim)
print('元素的个数', arr.size)
print('元素的数据类型', arr.dtype)
print('元素的转置\n', arr.T)
```

```python
arr = np.array([1, 2, 3])
print(arr)
print('数组的形状\n', arr.shape)
print('数组的维度', arr.ndim)
print('元素的个数', arr.size)
print('元素的数据类型', arr.dtype)
print('元素的转置\n', arr.T)
```

```python
arr = np.array([[1, 2, 3], [4, 5, 6]])
print(arr)
print('数组的形状\n', arr.shape)
print('数组的维度', arr.ndim)
print('元素的个数', arr.size)
print('元素的数据类型', arr.dtype)
print('元素的转置\n', arr.T)
```
