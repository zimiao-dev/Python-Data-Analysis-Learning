# ndarray
## 1. ndarray 的特性
### 多维性

```python
import numpy as np
```

```python
arr = np.array(5) # 创建一个 0 维 ndarray 数组
print(arr)
print('arr的维度: ', arr.ndim) # 数组的维度 number of dimensions
```

```python
arr = np.array([1, 2, 3]) # 创建一个 1 维的 ndarray 数组
print(arr)
print('arr的维度: ', arr.ndim) # 数组维度 number of dimensions
```

```python
arr = np.array([[1, 2, 3], [4, 5, 6]]) # 创建一个 2 维的 ndarray 数组
print(arr)
print('arr的维度: ', arr.ndim) # 数组维度 number of dimensions
```

### 同质性

```python
arr = np.array([1, 'hello']) # 数据类型不同，会被强制转换为相同的数据类型
print(arr)
```

```python
arr = np.array([1, 2.5]) # 数据类型不同，会被强制转换为相同的数据类型
print(arr)
```

## 2. ndarray 的属性

```python
arr = np.array(1)
print(arr)
print('数组的形状\n', arr.shape)
print('数组的维度', arr.ndim)
print('元素的个数', arr.size)
print('元素的数据类型', arr.dtype)
print('元素的转置\n', arr.T)
```

```python
arr = np.array([1, 2, 3])
print(arr)
print('数组的形状\n', arr.shape)
print('数组的维度', arr.ndim)
print('元素的个数', arr.size)
print('元素的数据类型', arr.dtype)
print('元素的转置\n', arr.T)
```

```python
arr = np.array([[1, 2, 3], [4, 5, 6]])
print(arr)
print('数组的形状\n', arr.shape)
print('数组的维度', arr.ndim)
print('元素的个数', arr.size)
print('元素的数据类型', arr.dtype)
print('元素的转置\n', arr.T)
```

## 3. ndarray 的创建

```python
# 基础创建方法
list1 = [4, 5, 6]
arr = np.array(list1, dtype=np.float64)
print(arr.ndim)
print(arr)
```

```python
# copy
arr1 = np.copy(arr) # 元素和原始数组相同，但已经是不同的数组了
print(arr1)
arr1[0] = 8
print(arr1)
print(arr)
```

```python
# 预定义形状
# 全 0 全 1 未初始化 固定值
# 全 0
arr = np.zeros((2, 3), dtype=int)
print(arr)
print(arr.dtype)
```

```python
arr = np.zeros((3, ), dtype=int)
print(arr)
print(arr.dtype)
```

```python
# 全 1
arr = np.ones((5, 8), dtype=int)
print(arr)
```

```python
# 未初始化
arr = np.empty((4, 5))
print(arr)
```

```python
# 固定值
arr = np.full((4,5), 315)
print(arr)
```

```python
arr1 = np.zeros_like(arr) # 形状一样
print(arr1)
arr1 = np.empty_like(arr) # 形状一样
print(arr1)
arr1 = np.ones_like(arr) # 形状一样
print(arr1)
arr1 = np.full_like(arr, 515) # 形状一样
print(arr1)
```

```python
# 等差数列 例如：2, 4, 6, 8
arr = np.arange(1, 51, 1) # start, end, step
print(arr)
```

```python
# 等间隔数列
arr = np.linspace(0, 100, 5)
print(arr)
```

```python
arr = np.linspace(0, 100, 5, dtype=int)
print(arr)
arr = np.arange(0, 101, 25)
print(arr)
```

```python
# 对数间隔数列
arr = np.logspace(0, 4, 3, base=2)
print(arr)
```

```python
arr = np.linspace(0, 4, 3)
print(arr)
```

```python
arr = np.logspace(0, 4, 3)
print(arr)
```

```python
# 特殊矩阵构造
# 单位矩阵：主对角线上的数字为 1 ，其他位置为 0
arr = np.eye(3, 4, dtype=int)
print(arr)
```

```python
# 对角矩阵：主对角线上是非 0 的数，其他位置为 0
arr = np.diag([1, 2, 3, 4, 5]) # 参数是主对角线上要填充的数字
print(arr)
```

```python
# 随机数组的生成
# 生成 0 - 1 之间的浮点数(均匀分布)
arr = np.random.rand(2, 3)
print(arr)
```

```python
# 生成指定范围区间的随机浮点数（均匀分布）
arr = np.random.uniform(3, 6, (2, 3))
print(arr)
```

```python
# 生成指定范围区间的随机整数（均匀分布）
arr = np.random.randint(3, 30, (2, 3))
print(arr)
```

```python
# 生成随机数列（正太分布）
# 两边的概率小，中间的概率大
arr = np.random.randn(2, 3)
print(arr)
```

```python
# 设置随机种子
np.random.seed(42)
arr = np.random.randint(1, 10, (2, 5))
print(arr)
```

## 05 索引与切片

```python
# 一维数组的索引与切片
arr = np.random.randint(1, 100, 20)
print(arr)
```

```python
print(arr[0])
print(arr[:]) # 获取全部数据
print(arr[2:5]) # start:end+1 左闭右开
print(arr[slice(2, 15, 3)]) # start, end, step
print(arr[(arr > 10) & (arr < 70)]) # 布尔索引
```

```python
# 二维数组的索引与切片
arr = np.random.randint(1, 100, (4, 8))
print(arr)
```

```python
print(arr[1, 3]) # 索引
print(arr[1, 2:5])
print(arr[arr > 50]) # 返回结果为一维数组
print(arr[2][arr[2] > 50]) # 第三行大于50
print(arr[:, 2][arr[:, 2] > 50]) # 第三列大于50
```
