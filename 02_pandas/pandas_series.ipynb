# Series
## 01 Series 的创建

```python
import pandas as pd
import numpy as np
```

```python
# series 的创建
s = pd.Series([1, 2, 3, 4, 5])
# 自定义索引
s = pd.Series([1, 2, 3, 4, 5], index=['A', 'B', 'C', 'D', 'E'])
# s = pd.Series([1, 2, 3, 4, 5], index=[1, 2, 3, 4, 5])
# 自定义 name
s = pd.Series([1, 2, 3, 4, 5], index=['A', 'B', 'C', 'D', 'E'], name='月份')
print(s)
```

```python
# 通过字典创建 Series
s = pd.Series({'a': 1, 'b':2, 'c': 3, 'd': 4, 'e': 5})
print(s)
s1 = pd.Series(s, index=['a', 'c'])
print(s1)
s = pd.Series([1, 2, 3, 4, 5], index=['A', 'B', 'C', 'D', 'E'], name='月份')
s1 = pd.Series(s, index=['A', 'C'])
print(s1)
```

## 02 Series 的属性

```python
# Series 的属性
'''
index: Series 的索引
values: Series 的值
dtype/dtypes: Series 的元素类型
shape: Series 的形状
ndim: Series 的维度
size: Series 的元素个数
name: Series 的名称
loc[]: 显示索引，按照标签索引或切片
iloc[]: 隐式索引，按照位置索引或切片
at[]: 使用标签访问单个元素
iat[]: 使用位置访问单个元素
'''
```

```python
s = pd.Series({'a': 1, 'b':2, 'c': 3, 'd': 4, 'e': 5})
print(s.index)
print(s.values)
print(s.shape)
print(s.ndim)
print(s.size)
print(s.dtype)
print(s.name)
print(s.loc['b']) # 显示索引
print(s.loc['a':'c':2])
print(s.iloc[1]) # 隐式索引
print(s.iloc[0:3:2])
print(s.at['a'])
print(s.iat[0])
```

## 03 Series 访问数据

```python
# Series 访问数据方式
# print(s[1])
print(s['a'])
print(s)
print(s[s < 3])
s['f'] = 6
print(s.head()) # 前几行（默认5）
print(s.tail()) # 末几行（默认5）
print(s.head(2))
print(s.tail(2))
```

## Series 常用方法

```python
# Series 常用函数
s = pd.Series([10, 2, np.nan, None, 3, 4, 5], ['A', 'B', 'C', 'D', 'E', 'F', 'G'], name='data')
print(s)
```

```python
s.head(3) # 默认取前 5 行数据
s.tail(6) # 默认取后 5 行数据
```

```python
# 查看所有的描述性信息
print(s.describe())
```

```python
# 获取元素个数（忽略缺失值）
print(s.count())
```

```python
# 获取索引
print(s.keys()) # 方法
print(s.index) # 属性
```

```python
print(s.isna()) # 检查Series 的每个元素是否为为缺失值
```

```python
s.isin([4, 5, 6]) # 检查每个元素是否在集合中
```

```python
print(s.describe())
```

```python
print(s.mean()) # 平均值
print(s.sum()) # 和
print(s.std()) # 标准差
print(s.var()) # 方差
print(s.min()) # 最小值
print(s.median()) # 中位数
print(s.max()) # 最大值
```

```python
s.sort_values()
print(s.quantile(0.25))
print(s.quantile(0.75))
print(s.quantile(0.8)) # 分位数
# --------------
# 2  3  4  5  10
# 位置 4 * 0.8 = 3.2
# 值 5 + (10 - 5) * 0.2 = 6
```

```python
# 众数
print(s.mode())
s['H'] = 4
print(s.mode())
```

```python
print(s.value_counts()) # 每个元素的计数
```

```python
s.drop_duplicates() # 去重
s.unique()
print(s.nunique()) # 去重后的元素个数
```

```python
# 排序：按值、索引
s.sort_index()
s.sort_values()
```
