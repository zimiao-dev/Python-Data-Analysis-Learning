## 案例1 学生成绩统计
创建一个包含 10 名学生成绩的 Series , 成绩范围在 50 - 100 之间。

计算平均分、最高分、最低分，找出高于平均分的学生人数。


```python
# 生成随机成绩
np.random.seed(42)
# scores = pd.Series(np.random.randint(50, 101, 10), index=['学生' + str(i) for i in range(1, 11)])
values = np.random.randint(50, 101, 10)
# indexes = ['学生' + str(i) for i in range(1, 11)]
indexes = []
for i in range(1, 11):
    indexes.append('学生' + str(i))
scores = pd.Series(values, index=indexes)
# print(scores)
print('平均分:', scores.mean())
print('最高分:', scores.max())
print('最低分:', scores.min())
print('高于平均分的学生人数:', scores[scores > scores.mean()].size)
print('高于平均分的学生人数:', len(scores[scores > scores.mean()]))
print('高于平均分的学生人数:', scores[scores > scores.mean()].count())
```

## 案例 2 ：温度数据分析
给定某城市一周每天的最高气温 Series，完成以下任务：

- 找出温度超过 30 度的天数
- 计算平均气温
- 将温度从高到低排列
- 找出温度变化最大的两天

```python
temperatures = pd.Series([28, 31, 29, 32, 30, 27, 33], index=['周一', '周二', '周三', '周四', '周五', '周六', '周日'])
print('温度高于 30 的天数: ', temperatures[temperatures > 30].count())
print('平均气温：', temperatures.mean())
print('温度从高到底排列：')
print(temperatures.sort_values(ascending=False))
```

```python
# 找出温度变化最大的两天
# 28   31   29   32   30   27   33
# None  3   -2    3   -2   -3    6
print('温度变化最大的两天为:', *temperatures.diff().abs().sort_values(ascending=False).keys()[:2].to_list())
```

## 案例 3 ：股票价格分析
给定某只股票连续 10 个交易日的收盘价 Series

- 计算每日收益率（当日收盘价/前日收盘价 - 1）
- 找出收益率最高和最低的日期
- 计算波动率（收益率的标准差）

```python
prices = pd.Series([102.3, 103.5, 105.1, 104.8, 106.2, 107.0, 106.5, 108.1, 109.3, 110.2], index=pd.date_range('2023-01-01', periods=10))
# print(prices.diff()/prices)
# print('收益率最高的日期：', (prices.diff()/prices).sort_values(ascending=False).keys()[0])
# print('收益率最高的日期：', (prices.diff()/prices).sort_values().keys()[0])
# print('波动率为：', (prices.diff()/prices).std())
# 计算每日收益率
a = prices.pct_change()
print(a)
# 收益率最高日期
print(a.idxmax())
# 收益率最低日期
print(a.idxmin())
# 波动率
print(a.std())
```

##  案例 4：销售数据分析
某产品过去 12 个月的销售量 Series

- 计算季度平均销量（每三个月为一个季度）
- 找出销售量最高的月份
- 计算月环比增长率
- 找出连续增长超过两个月的月份


```python
sales = pd.Series([120, 135, 145, 160, 155, 170, 180, 175, 190, 200, 210, 220], index=pd.date_range('2024-01-01', periods=12, freq='MS'))
print(sales)
```

```python
# 季度的平均销量
sales.resample('QS').mean() # 重新采样
```

```python
print('销量最高的月份：', sales.idxmax())
```

```python
# 月环比增长率
sales.pct_change()
```

```python
b = sales.pct_change() > 0
b[b.rolling(3).sum()==3].keys().tolist()
```

## 案例 5 ：每小时销售分析
某商店每小时销售额Series

- 按天重采样计算总销售额
- 计算每天营业时间(8:00 - 20:00)和非营业时间的销售额比例
- 找出销售额最高的 3 小时

```python
import pandas as pd
import numpy as np
np.random.seed(42)
hours_sales = pd.Series(np.random.randint(0, 100, 24), index=pd.date_range('2025-01-01', periods=24, freq='h'))
print(hours_sales)
```

```python
day_sales = hours_sales.resample('D').sum()
print(day_sales)
# hours_sales.sum()
```

```python
hours_sales.between_time('8:00', '22:00').sum() # 筛选一段时间内的 Series
business_hours_sales = hours_sales[(hours_sales.index.hour >= 8) & (hours_sales.index.hour <= 22)]
# print(business_hours_sales.sum()/(day_sales - business_hours_sales.sum()))
# not_business_hours_sales = hours_sales.drop(business_hours_sales.index)
# not_business_hours_sales = hours_sales[(hours_sales.index.hour < 8) | (hours_sales.index.hour > 22)]
mask = (hours_sales.index.hour >= 8) & (hours_sales.index.hour <= 22)
business_hours_sales = hours_sales[mask]
not_business_hours_sales = hours_sales[~mask]
print(business_hours_sales.sum() / not_business_hours_sales.sum())
```

```python
# 找出销售额最高的3个小时
print(hours_sales.nlargest(3).index)
print(hours_sales.nlargest(3).keys())
print(hours_sales.nlargest(3).values)
```
