```python
import numpy as np
import pandas as pd
```

```python
s1 = pd.Series([1, 2, 3, 4, 5])
s2 = pd.Series([6, 7, 8, 9, 10])
df = pd.DataFrame({'第一列': s1, '第二列': s2})
# print(type(df['第一列']))
# 通过字典创建
df = pd.DataFrame(
    {
       'name': ['tom', 'jack', 'alice', 'bob', 'allen'],
        'age': [15, 17, 20, 26, 30],
        'score': [60.5, 80, 30.6, 70, 83.6]
    }, index=[1, 2, 3, 4, 5], columns=['name', 'age', 'score']
)
df
```
