# DataFrame
## 01 DataFrame 的创建方式

```python
import numpy as np
import pandas as pd
from pandas.core.algorithms import nunique_ints
```

```python
s1 = pd.Series([1, 2, 3, 4, 5])
s2 = pd.Series([6, 7, 8, 9, 10])
df = pd.DataFrame({'第一列': s1, '第二列': s2})
# print(type(df['第一列']))
# 通过字典创建
df = pd.DataFrame(
    {
       'name': ['tom', 'jack', 'alice', 'bob', 'allen'],
        'age': [15, 17, 20, 26, 30],
        'score': [60.5, 80, 30.6, 70, 83.6]
    }, index=[1, 2, 3, 4, 5], columns=['name', 'score', 'age']
)
df
```

## 02 DataFrame 的属性

```python
# DataFrame 的属性
print('行索引：')
print(df.index)
print('列标签：')
print(df.columns)
print('值：')
print(df.values)
```

```python
print('维度：', df.ndim)
print('形状：', df.shape)
print('元素个数：', df.size)
print('数据类型：')
print(df.dtypes)
```

```python
# 行列转置
print('行列转置：')
print(df.T)
print(df.T.index)
print(df.T.columns)
```

```python
print(df)
```

```python
# 获取元素 loc iloc at iat
# 获取某行
print(df.loc[4, :])
print(df.iloc[3, :])

# 获取某列
print(df.loc[:, 'name'])
print(df.iloc[:, 0])
```

```python
# 获取单个数据
print(df.at[3, 'score'])
print(df.loc[3, 'score'])
print(df.iat[2, 1])
print(df.iloc[2, 1])
```

## 03 DataFrame 数据访问

```python
# 获取当列数据, df[列标签]
print(df['name'])
print(type(df['name']))
print(df.name)
print(type(df.name))
print(df[['name']])
print(type(df[['name']]))
df[['name', 'score']]
```

```python
print(df[['name', 'score']]) # 多列数据的获取，返回 DataFrame
```

```python
# 查看部分数据
print(df.head(3))
```

```python
print(df.tail(2))
```

```python
# 使用布尔索引筛选数据
print(df[df.score > 70])
print(df[(df.age > 20) & (df.score > 70.0)])
df[(df['age'] > 20) & (df['score'] > 70.0)]
```

```python
# 随机取样
df.sample(3)
```

## 04 DataFrame 的常用方法

```python
print(df.head()) # 查看前 n 行数据，默认前 5 行
print(df.tail(1)) # 查看后 n 行数据，默认后 5 行
```

```python
print(df.isin(['jack'])) # 查看每个元素是否在参数集合中
print(df.isin(['jack', 20]))
```

```python
print(df.isna()) # 查看每个元素是否是缺失值
```

```python
print(df['age'].sum()) # 求某一列的总和
print(df['score'].max()) # 最大值
print(df['age'].min()) # 最小值
print(df['score'].mean()) # 平均值
print(df['score'].median()) # 中位数
print(df['score'].mode()) # 众数
print(df['age'].mode())
```

```python
df = pd.DataFrame(
    {
       'name': ['tom', 'jack', 'alice', 'bob', 'allen'],
        'age': [15, 15, 20, 26, 30],
        'score': [60.5, 80, 30.6, 70, 83.6]
    }, index=[1, 2, 3, 4, 5], columns=['name', 'score', 'age']
)
print(df['age'].mode()) # 众数
```

```python
print(df.score.std()) # 标准差
print(df.score.var()) # 方差
print(df.score.quantile(0.25))
```

```python
print(df.describe())
```

```python
print(df.count())
```

```python
df = pd.DataFrame(
    {
       'name': ['tom', 'tom','jack', 'alice', 'bob', 'allen'],
        'age': [15, 15, 15, 20, 26, 30],
        'score': [60.5, 60.5,  80, 30.6, 70, 83.6]
    }, index=[1, 2, 3, 4, 5, 6], columns=['name', 'age', 'score']
)
print(df.value_counts()) # 每条记录出现的次数
print(df.nunique())
```

```python
print(df.drop_duplicates())
```

```python
print(df.duplicated())
print(df.duplicated(subset=['age'])) # 查看是否重复
```

```python
df.sample(3) # 随机抽样
```

```python
print(df.replace(15, 30))
```

```python
# 累计和 累积积
df
```

```python
df.cumsum() # 累积和
df.cumsum(axis=0) # 默认 axis=0 表示按列累积
```

```python
df.cummax()
```

```python
df.cummin()
```

```python
print(df.sort_index(ascending=False)) # 按索引从大到小排序
```

```python
print(df.sort_values(by='name')) # 按照列排序
```

```python
df = pd.DataFrame(
    {
       'name': ['tom', 'tom','jack', 'alice', 'bob', 'allen'],
        'age': [15, 15, 15, 20, 26, 30],
        'score': [60.5, 60.5,  80, 30.6, 70, 80]
    }, index=[1, 2, 3, 4, 5, 6], columns=['name', 'score', 'age']
)
```

```python
print(df.sort_values(by='score'))
```

```python
print(df.sort_values(by=['score', 'age'], ascending=[True, False])) # 按照多列排序
```

```python
print(df.nlargest(2, columns=['score', 'age'])) # 最大几个值
```

```python
print(df.nsmallest(2, columns=['score', 'age']))
```
