# DataFrame
## 01 DataFrame 的创建方式

```python
import numpy as np
import pandas as pd
```

```python
s1 = pd.Series([1, 2, 3, 4, 5])
s2 = pd.Series([6, 7, 8, 9, 10])
df = pd.DataFrame({'第一列': s1, '第二列': s2})
# print(type(df['第一列']))
# 通过字典创建
df = pd.DataFrame(
    {
       'name': ['tom', 'jack', 'alice', 'bob', 'allen'],
        'age': [15, 17, 20, 26, 30],
        'score': [60.5, 80, 30.6, 70, 83.6]
    }, index=[1, 2, 3, 4, 5], columns=['name', 'score', 'age']
)
df
```

## 02 DataFrame 的属性

```python
# DataFrame 的属性
print('行索引：')
print(df.index)
print('列标签：')
print(df.columns)
print('值：')
print(df.values)
```

```python
print('维度：', df.ndim)
print('形状：', df.shape)
print('元素个数：', df.size)
print('数据类型：')
print(df.dtypes)
```

```python
# 行列转置
print('行列转置：')
print(df.T)
print(df.T.index)
print(df.T.columns)
```

```python
print(df)
```

```python
# 获取元素 loc iloc at iat
# 获取某行
print(df.loc[4, :])
print(df.iloc[3, :])

# 获取某列
print(df.loc[:, 'name'])
print(df.iloc[:, 0])
```

```python
# 获取单个数据
print(df.at[3, 'score'])
print(df.loc[3, 'score'])
print(df.iat[2, 1])
print(df.iloc[2, 1])
```
