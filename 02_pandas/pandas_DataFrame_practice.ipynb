## 案例 1 ：学生成绩分析
场景：某班级的学生成绩如下，请完成如下任务：

- 计算每位学生的总分和平均分
- 找出数学成绩高于 90 分或英语成绩高于 85 分的学生
- 按总分从高到低排序，并输出前三名学生

```python
import pandas as pd
data = {
        '姓名': ['张三', '李四', '王五', '赵六', '钱七'],
        '数学': [82, 92, 78, 88, 95],
        '英语': [98, 88, 85, 92, 80],
        '物理': [75, 80, 88, 85, 90]
    }
```

```python
scores = pd.DataFrame(data)
scores
```

```python
scores['总分'] = scores[['数学', '英语', '物理']].sum(axis=1)
# scores['平均分'] = scores['总分'] / 3
scores['平均分'] = scores[['数学', '英语', '物理']].mean(axis=1)
scores
```

```python
scores[(scores['数学'] > 90) | (scores['英语'] > 85)]
```

```python
print(scores.sort_values(by='总分', ascending=False).head(3))
print(scores.nlargest(3, columns='总分'))
```

## 案例 2： 销售数据分析
场景：某公司的销售数据如下，请完成如下任务：

- 计算每种产品的总销售额（销售额 = 单价 * 销量）
- 找出销售额最高的产品
- 按销售额从高到低排序，输出所有产品信息

```python
import pandas as pd
data = {
    '产品名称': ['A', 'B', 'C', 'D'],
    '单价': [100, 150, 200, 120],
    '销量': [50, 30, 20, 40]
}
```

```python
df = pd.DataFrame(data)
df
```

```python
df['销售额'] =  df['单价'] * df['销量']
df
```

```python
df.nlargest(1, columns='销售额')
```

```python
df.sort_values(by=['销售额'], ascending=False)
```

## 案例 3 ：电商用户行为分析
某电商的用户行为数据如下， 请完成以下任务：

- 计算每位用户的总消费金额 (消费金额 = 商品单价 * 购买数量)
- 找出消费金额最高的用户并输出所有信息
- 计算所有用户的平均消费金额（保留两位小数）
- 统计电子产品的总购买数量

```python
import pandas as pd
data = {
    '用户ID': [101, 102, 103, 104, 105],
    '用户名': ['Alice', 'Bob', 'Charlie', 'David', 'Eve'],
    '商品类别': ['电子产品', '服饰', '电子产品', '家居', '服饰'],
    '商品单价': [1200, 300, 800, 150, 200],
    '购买数量': [1, 3, 2, 5, 4]
}
```

```python
df = pd.DataFrame(data)
df
```

```python
df['总消费金额'] = df['商品单价'] * df['购买数量']
df
```

```python
df.nlargest(1, columns='总消费金额')
```

```python
print('平均消费金额：%.2f' % df['总消费金额'].mean())
```

```python
print(df[df['商品类别'] == '电子产品']['购买数量'].sum())
```
