### 03 重复数据处理

```python
import pandas as pd

data = {
    'name': ['alice', 'alice', 'bob', 'alice', 'jack', 'bob'],
    'age': [26, 25, 30, 25, 35, 30],
    'city': ['NY', 'NY', 'LA', 'NY', 'SF', 'LA']
}
df = pd.DataFrame(data)
```

```python
df
```

```python
df.duplicated() # 整条记录都相同，标记为重复，True
```

```python
df.drop_duplicates()
```

```python
df.drop_duplicates(subset='name') # 根据指定列去重
```

```python
df.drop_duplicates(subset='name', keep='last') # 根据指定列去重，保留最新出现的行
```

## 06 数据分箱

```python
# 数据分箱 pd.cut(x, bins, labels)
import pandas as pd
df = pd.read_csv('../data/employees.csv')
```

```python
df.head(10)
```

```python
df1 = df[['employee_id', 'salary']].head(10) # 先取前10条调试
```

```python
df1
```

```python
pd.cut(df1['salary'], bins=3) # bins=n, 分成n段区间，起始值和结束值是所有数据的最大值和最小值
# 4180 ~ 14100 ~ 24000
```

```python
pd.cut(df1['salary'], bins=3).value_counts()
```

```python
pd.cut(df1['salary'], bins=[0, 10000, 20000, 30000]) # bins为list时，分为指定的区间
```

```python
pd.cut(df1['salary'], bins=[0, 10000, 20000, 30000]).value_counts()
```

```python
pd.cut(df1['salary'], bins=[0, 10000, 20000, 30000], labels=['低', '中', '高']) # labels 指定每个区间的名称
```

```python
pd.cut(df1['salary'], bins=[0, 10000, 20000, 30000], labels=['低', '中', '高']).value_counts()
```

```python
df1['收入水平'] = pd.cut(df1['salary'], bins=[0, 10000, 20000, 30000], labels=['低', '中', '高'])
```

```python
print(df1)
```

```python
pd.qcut(df1['salary'], 3) # 按照频率分箱
```

```python
pd.qcut(df1['salary'], 3).value_counts()
```
