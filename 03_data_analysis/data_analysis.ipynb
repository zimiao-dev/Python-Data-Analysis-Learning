## 02 缺失值处理

```python
import pandas as pd
import numpy as np
s = pd.Series([15, 21, np.nan, None, pd.NA])
s
```

```python
print(s.isna())
print(s.isnull())
```

```python
# 查看确实值个数
s.isna().sum()
```

```python
df = pd.DataFrame([[1, pd.NA, 2], [2, 3, 5], [None, 4, 6]])
df.columns = ['第一列', '第二列', '第三列']
df
```

```python
# 查看是否是缺失值
df.isna()
```

```python
df.isnull()
```

```python
 # 查看缺失值个数
df.isna().sum(axis=1)
```

```python
df.isna().sum()
```

```python
# 剔除缺失值
s.dropna()
```

```python
df.dropna() # 含缺失值的记录被整条删除
```

```python
df.dropna(how='all') # 如果整行都是缺失值，就保留
```

```python
df.dropna(thresh=2) # 如果至少有 n 个不是缺失值，就保留
```

```python
df.dropna(axis=1) # 剔除一整列
```

```python
df.dropna(subset=['第一列']) # 剔除指定列中有缺失值的行
```

```python
# 填充缺失值
df = pd.read_csv('../data/weather_withna.csv')
df.tail()
```

```python
df.isna().sum(axis=0)
```

```python
df.head()
```

```python
# 填充缺失值
df.fillna({'temp_max': 20}).tail() # 使用字典填充
```

## 03 重复数据处理

```python
import pandas as pd
import numpy as np

data = {
    'name': ['alice', 'alice', 'bob', 'alice', 'jack', 'bob'],
    'age': [26, 25, 30, 25, 35, 30],
    'city': ['NY', 'NY', 'LA', 'NY', 'SF', 'LA']
}
df = pd.DataFrame(data)
```

```python
df
```

```python
df.duplicated() # 整条记录都相同，标记为重复，True
```

```python
df.drop_duplicates()
```

```python
df.drop_duplicates(subset='name') # 根据指定列去重
```

```python
df.drop_duplicates(subset='name', keep='last') # 根据指定列去重，保留最新出现的行
```

## 05 数据变形

```python
import pandas as pd
```

```python
data = {
    'ID': [1, 2],
    'name': ['alice', 'bob'],
    'math': [90, 85],
    'English': [88, 92],
    'Science': [95, 89]
}
```

```python
df = pd.DataFrame(data)
```

```python
df
```

```python
df.T # 行列转置
```

```python
# 宽表转长表
df2 = pd.melt(df, id_vars=['ID', 'name'], var_name='科目', value_name='分数').sort_values('name')
```

```python
df2
```

```python
# 长表转宽表
df3 = pd.pivot(df2, index=['ID', 'name'], columns=['科目'], values=['分数'])
```

```python
df3
```
