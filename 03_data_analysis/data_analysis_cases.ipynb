## 01 企鹅数据分析

```python
# 一般步骤：
    # 1. 导入一些必要的库
    # 2. 导入数据
    # 3. 数据清洗
    # 4. 数据特征的构造
    # 5. 数据分析
```

```python
# 1. 导入一些必要的库
import pandas as pd
```

```python
# 2. 导入数据
df = pd.read_csv('../data/penguins.csv')
```

```python
df.head(5)
```

```python
df.info()
```

```python
# 3. 数据清洗
df.isna().sum(axis=0)
```

```python
df.dropna(inplace=True)
```

```python
df.isna().sum()
```

```python
df.info()
```

```python
# 4. 数据特征的构造
df.head(5)
```

```python
df['sex'] = df['sex'].astype('category')
```

```python
df['bill_radio'] = df['bill_length_mm'] / df['bill_depth_mm']
```

```python
df.head(5)
```

```python
# 5. 数据分析
# 数据分箱：体重分为3个等级
labels = ['低', '中', '高']
df['body_level'] = pd.cut(df['body_mass_g'], bins=3, labels=labels)
```

```python
df['body_level'].value_counts()
```

```python
# 分组：按岛屿和性别分组分析
df.groupby(['sex'], observed=False).agg({
    'body_mass_g': ['mean', 'median', 'count']
})
```

```python
df.groupby(['island'], observed=False).agg({
    'body_mass_g': ['mean', 'median', 'count']
})
```
