### 01 时间数据处理

```python
import pandas as pd
# 时间数据处理
d = pd.Timestamp('2026-8-13 12:17') # 时间戳
```

```python
d
```

```python
type(d)
```

```python
# 时间戳可以取出的属性
print('年', d.year)
print('月', d.month)
print('日', d.day)
print('时', d.hour)
print('分', d.minute)
print('秒', d.second)
print('季度', d.quarter)
```

```python
# 判断相关
print('是否是月底', d.is_month_end)
```

```python
# 常用方法
print('星期', d.day_name()) # 星期几
print('转换为天维度', d.to_period('D')) # period 周期
print('转换为季度', d.to_period('Q'))
print('转换为年度', d.to_period('Y'))
print('转换为月度', d.to_period('M'))
print('转换为周维度', d.to_period('W'))
```

```python
# 字符串转为日期类型
d = pd.to_datetime('2026/08/13 12:37:15')
```

```python
# DataFrame 中的日期转换

df = pd.DataFrame({
    'sales': [100, 200, 300],
    'date': ['20260810', '20260811', '20260812']
})
```

```python
df
```

```python
df['datetime'] = pd.to_datetime(df['date'])
```

```python
df.info()
```

```python
df['week'] = df['datetime'].dt.day_name() # DateFrame中使用datetime的属性和方法，加一个.dt
```
