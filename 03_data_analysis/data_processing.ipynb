### 01 时间数据处理

```python
import pandas as pd
# 时间数据处理
d = pd.Timestamp('2026-8-13 12:17') # 时间戳
```

```python
d
```

```python
type(d)
```

```python
# 时间戳可以取出的属性
print('年', d.year)
print('月', d.month)
print('日', d.day)
print('时', d.hour)
print('分', d.minute)
print('秒', d.second)
print('季度', d.quarter)
```

```python
# 判断相关
print('是否是月底', d.is_month_end)
```

```python
# 常用方法
print('星期', d.day_name()) # 星期几
print('转换为天维度', d.to_period('D')) # period 周期
print('转换为季度', d.to_period('Q'))
print('转换为年度', d.to_period('Y'))
print('转换为月度', d.to_period('M'))
print('转换为周维度', d.to_period('W'))
```

### 02 分组聚合

```python
# df.groupby('分组的字段')['聚合的字段'].聚合函数()
```

```python
import pandas as pd
```

```python
df = pd.read_csv('../data/employees.csv')
```

```python
df
```

```python
df.info()
```

```python
df['department_id'].isna().sum()
```

```python
df.dropna(subset=['department_id'], inplace=True)
df['department_id'].isna().sum()
```

```python
df['department_id'] = df['department_id'].astype('int')
```

```python
# 计算不同部门的平均薪资
```

```python
df.groupby('department_id')
```

```python
df.groupby('department_id').groups # 查看所有的分组
```

```python
df.groupby('department_id').get_group(30) # 查看具体的某个分组数据
```

```python
df.groupby('department_id')[['salary']]
```

```python
df.groupby('department_id')[['salary']].mean()
```

```python
df2 = df.groupby('department_id')[['salary']].mean()
```

```python
df2['salary'] = df2['salary'].round(2)
```

```python
df2
```

```python
df2 = df2.reset_index()
```

```python
df2
```

```python
df2.sort_values('salary', ascending=False)
```

```python
# 计算不同部门不同岗位的平均薪资情况
```

```python
df.groupby(['department_id', 'job_id'])
```

```python
df.groupby(['department_id', 'job_id']).groups
```

```python
df_new = df.groupby(['department_id', 'job_id'])[['salary']].mean()[['salary']].round(1)
```

```python
df_new.reset_index(inplace=True)
```

```python
df_new.sort_values('salary', ascending=False)
```
