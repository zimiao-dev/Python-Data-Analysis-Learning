## 01 折线图

```python
import matplotlib.pyplot as plt
from matplotlib import rcParams
# rcParams['font.family'] = 'SimHei' # 设置字体
rcParams['font.sans-serif'] = 'SimHei'
```

```python
# 创建图表，设置大小
plt.figure(figsize=(10, 5))

# 绘图数据
month = ['1月', '2月', '3月', '4月']
sales = [100, 150, 80, 130]

# 绘制图表
plt.plot(month, sales, label='产品A', color='orange', linewidth=2, linestyle='--', marker='o') # 这里的label表示图例

# 添加标题
plt.title('2026年销售趋势', color='red', fontsize=20)

# 添加坐标轴标签
plt.xlabel('月份', fontsize=10)
plt.ylabel('销售额(万元)', fontsize=10)

# 添加图例
plt.legend(loc='upper left')

# 添加网格线
# plt.grid(True) # 横竖线
# plt.grid(axis='x') # 竖线
# plt.grid(axis='y') # 横线
plt.grid(True, alpha=0.3, color='white', linestyle='--') # linestyle : '-' 实线，'--'虚线

# 设置轴刻度的字体大小
plt.xticks(rotation=0, fontsize=12)
plt.yticks(rotation=0, fontsize=12)

# 设置 y 的范围
plt.ylim(70, 170)

for x, y in zip(month, sales):
    plt.text(x, y + 2, str(y), ha='center', va='bottom', fontsize=10)

# 显示图表
plt.show()
```

## 02 柱状图和条形图

```python
import matplotlib.pyplot as plt
from matplotlib import rcParams

# rcParams['font.family'] = 'SimHei' # 设置字体
rcParams['font.sans-serif'] = 'SimHei'
```

```python
# 设置图表画布大小
plt.figure(figsize=(10, 5))

# 绘图数据
subjects = ['语文', '数学', '英语', '科学']
scores = [85, 92, 78, 88]

# 绘制柱状图
plt.bar(subjects, scores, label='小红', width=0.4)

# 设置标题
plt.title('小红2026年成绩分布', color='red', fontsize=20)

# 设置坐标轴标签
plt.xlabel('科目', fontsize=10)
plt.ylabel('科目', fontsize=10)

# 添加图例
plt.legend(loc='upper left')

# 添加网格线
plt.grid(axis='y', alpha=0.3, color='blue', linestyle='--')

# 设置刻度字体
plt.xticks(rotation=0, fontsize=12)
plt.yticks(rotation=0, fontsize=12)

# 设置 y 轴范围
plt.ylim(0, 120)

# 每个数据点上显示数值
for x, y in zip(subjects, scores):
    plt.text(x, y + 2, str(y), ha='center', va='bottom', fontsize=10)

# 自动优化排版
plt.tight_layout()

# 显示图表
plt.show()
```

```python
# 设置图表画布大小
plt.figure(figsize=(10, 5))

# 绘图数据
countries = ['United States', 'China', 'Japan', 'Germany', 'India']
gdp = [92, 78, 43, 22, 16]

# 绘制条形图
plt.barh(countries, gdp, height=0.5)

# 设置标题
plt.title('20XX年GDP排名', color='red', fontsize=20)

# 设置坐标轴标签
plt.xlabel('GDP', fontsize=10)
plt.ylabel('国家', fontsize=10)

# 自动化排版
plt.tight_layout()

# 显示图表
plt.show()
```

## 03 饼图 和 环形图

```python
import matplotlib.pyplot as plt
from matplotlib import rcParams

# rcParams['font.family'] = 'SimHei' # 设置字体
rcParams['font.sans-serif'] = 'SimHei'
```

```python
# 设置图表画布大小
plt.figure(figsize=(10, 5))

# 绘图数据
things = ['学习', '娱乐', '运动', '睡觉', '其他']
times =[6, 4, 1, 8, 5]

# 设置配色
colors = ['#66b3ff', '#99ff99', '#ffcc99', '#ff9999', '#ff4499']

# 绘制饼图
plt.pie(times, labels=things,
        autopct='%1.1f%%', # 显示百分比
        startangle=90, # startangle 调整初始画图的角度
        colors=colors
        )

# 设置标题
plt.title('一天的时间分布', color='red', fontsize=20)

# 自动优化图表
plt.tight_layout()

# 显示图表
plt.show()
```

```python
# 设置图表画布大小
plt.figure(figsize=(10, 5))

# 绘图数据
things = ['学习', '娱乐', '运动', '睡觉', '其他']
times =[6, 4, 1, 8, 5]

# 设置配色
colors = ['#66b3ff', '#99ff99', '#ffcc99', '#ff9999', '#ff4499']

# 绘制环形图
plt.pie(times, labels=things,
        autopct='%1.1f%%', # 显示百分比
        startangle=90, # startangle 调整初始画图的角度
        colors=colors,
        wedgeprops={'width': 0.6}, # 设置饼图宽度
        pctdistance=0.8 # 设置百分比位置
        )

# 设置标题
plt.title('一天的时间分布', color='red', fontsize=20)

# 写文字
plt.text(0, 0, '总计 100%', ha='center', va='bottom', fontsize=10)

# 自动优化图表
plt.tight_layout()

# 显示图表
plt.show()
```

```python
# 设置图表画布大小
plt.figure(figsize=(10, 5))

# 绘图数据
things = ['学习', '娱乐', '运动', '睡觉', '其他']
times =[6, 4, 1, 8, 5]

# 设置配色
colors = ['#66b3ff', '#99ff99', '#ffcc99', '#ff9999', '#ff4499']

# 设置突出块位置
explode = [0.1, 0, 0, 0, 0] # 设置每个块突出的程度

# 绘制饼图
plt.pie(times, labels=things,
        autopct='%1.1f%%', # 显示百分比
        startangle=0, # startangle 调整初始画图的角度
        colors=colors,
        explode=explode, # 设置突出块
        shadow=True # 设置阴影
        )

# 设置标题
plt.title('一天的时间分布', color='red', fontsize=20)

# 自动优化图表
plt.tight_layout()

# 显示图表
plt.show()
```
