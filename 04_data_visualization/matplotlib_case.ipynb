## 案例：温度分析

```python
# 导入所需的库
import pandas as pd
import matplotlib.pyplot as plt

# 设置中文字体
plt.rcParams['font.sans-serif'] = ['SimHei']

# 解决负号显示问题
plt.rcParams['axes.unicode_minus'] = False
# print(plt.rcParams)
```

```python
# 导入数据
df = pd.read_csv('../data/weather.csv')
display(df.head())
```

1. 绘制气温趋势变化图

```python
df['date'] = pd.to_datetime(df['date'])

# 筛选出 2015 年的数据，并创建独立副本
df_2015 = df.loc[df['date'].dt.year == 2015].copy()
# df_2015.info()

plt.figure(figsize=(10, 5))

plt.plot(df_2015['date'], df_2015['temp_max'], label='最高气温')
plt.plot(df_2015['date'], df_2015['temp_min'], label='最低气温')

plt.title('2015 年气温趋势变化图', fontsize=20)

plt.xlabel('日期', fontsize=10)
plt.ylabel('气温', fontsize=10)

plt.legend(loc='upper right')
```

```python
df_2015['temp_mean'] = (df['temp_max'] + df['temp_min']) / 2
# display(df_2015)

plt.figure(figsize=(10, 5))

plt.plot(df_2015['date'], df_2015['temp_mean'], label='平均气温')

plt.title('2015 年气温趋势变化图', fontsize=20)

plt.xlabel('日期', fontsize=10)
plt.ylabel('气温', fontsize=10)

plt.legend(loc='upper right')
```
